Kalkulator Faktorial

1. Meminta pengguna memasukkan sebuah bilangan bulat antara 1 sampai 5.
2. Menghitung faktorial dari bilangan tersebut menggunakan if / elif / else — bukan dengan fungsi bawaan Python.
3. Menampilkan hasil faktorial.
4. Jika input di luar rentang 1–5, tampilkan pesan bahwa input tidak valid.

```python

n = int(input("Masukkan bilangan bulat 1-5: "))

if n < 1 or n > 5:
    print("Input tidak valid")
elif n == 1:
    hasil = 1
elif n == 2:
    hasil = 2 * 1
elif n == 3:
    hasil = 3 * 2 * 1
elif n == 4:
    hasil = 4 * 3 * 2 * 1
else:
    hasil = 5 * 4 * 3 * 2 * 1

if 1 <= n <= 5:
    print("Hasil faktorial =", hasil)
```

Klasifikasi Kualitas Daya Berdasarkan Faktor Daya dan Efisiensi Distribusi

1. Meminta faktor daya (nilai antara 0.0 hingga 1.0) dan efisiensi distribusi (%).
2. Menghitung skor kualitas daya dengan rumus: skor = faktor_daya * efisiensi.
3. Mengklasifikasikan kualitas daya berdasarkan tabel berikut:

      1. skor >= 85 dan faktor_daya >= 0.90	PRIMA — sistem dalam kondisi optimal
      
      2. skor >= 70 dan faktor_daya >= 0.80	BAIK — sistem memadai, pantau ruti
      
      3. skor >= 50 atau faktor_daya >= 0.75	CUKUP — ada ruang perbaikan
      
      4. selain itu	BURUK — perlu perbaikan segera

```python

faktor_daya = float(input("Masukkan faktor daya (0-1): "))
efisiensi = float(input("Masukkan efisiensi (%): "))

skor = faktor_daya * efisiensi

if skor >= 85 and faktor_daya >= 0.90:
    print("PRIMA")
elif skor >= 70 and faktor_daya >= 0.80:
    print("BAIK")
elif skor >= 50 or faktor_daya >= 0.75:
    print("CUKUP")
else:
    print("BURUK")
```

Rekomendasi Mode Operasi PLTS Berdasarkan Iradiasi dan Kapasitas Baterai

1. Meminta iradiasi matahari (W/m²) dan SoC baterai (%).
2. Menghitung estimasi daya panel dengan rumus: daya_panel = iradiasi * 0.18 * 10 (asumsi: luas panel 10 m², efisiensi 18%).
3. Menentukan mode operasi berdasarkan tabel berikut:

Kondisi	Mode Operasi

  1. iradiasi > 600 dan SoC < 80	CHARGING — iradiasi tinggi, baterai belum penuh, prioritaskan pengisian
    
  2. iradiasi > 600 dan SoC >= 80	GRID-EXPORT — baterai sudah penuh, ekspor kelebihan daya ke jaringan
  
  3. iradiasi > 200 dan SoC >= 20	NORMAL — iradiasi dan baterai cukup, operasi mandiri
  
  4. iradiasi <= 200 dan SoC >= 50	BATTERY-ONLY — iradiasi rendah, andalkan baterai
  
  5. selain itu	EMERGENCY — iradiasi rendah dan baterai hampir habis, kurangi beban

```python

iradiasi = float(input("Masukkan iradiasi matahari (W/m2): "))
soc = float(input("Masukkan SoC baterai (%): "))

daya_panel = iradiasi * 0.18 * 10

if iradiasi > 600 and soc < 80:
    mode = "CHARGING"
elif iradiasi > 600 and soc >= 80:
    mode = "GRID-EXPORT"
elif iradiasi > 200 and soc >= 20:
    mode = "NORMAL"
elif iradiasi <= 200 and soc >= 50:
    mode = "BATTERY-ONLY"
else:
    mode = "EMERGENCY"

print("Estimasi daya panel:", daya_panel, "W")
print("Mode operasi:", mode)
```
