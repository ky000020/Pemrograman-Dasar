2. Energi Harian dan Capacity Factor PLTD

Sebuah PLTD beroperasi dengan daya_kw = 450 selama jam = 8 jam. Dalam sehari (24 jam) PLTD tersebut menghasilkan energi_aktual = 7200 kWh dari kapasitas_kw = 400.

a. Simpan nilai dalam variabel, hitung energi harian (energi = daya_kw × jam), lalu tampilkan dengan f-string.

b. Hitung capacity factor: Capacity factor (%) = (energi_aktual / (kapasitas_kw × 24)) × 100. Tampilkan dengan 2 angka desimal.

```python
#PLTD

daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# a. Simpan nilai dalam variabel, hitung energi harian (energi = daya_kw × jam),
# lalu tampilkan dengan f-string.

energi = daya_kw * jam

print(f"Energi Harian PLTD: {energi} kWh")


# b. Hitung capacity factor: Capacity factor (%) = (energi_aktual / (kapasitas_kw × 24)) × 100.
# Tampilkan dengan 2 angka desimal.

capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

print(f"Capacity Factor: {capacity_factor:.2f}%")
```

Pada program no 2 terdapat variabel daya_kw, jam, energi_aktual, energi, capacity factor dan kapasitas_kw yang membantu dalam penyimpanan data yang akan input, terdapat juga rumus yang digunakan untuk mencari jumlah energi yang digunakan yaitu energi = daya_kw * jam hasil dari rumus tersbut akan di simpan didalam variabel energi.

Kita juga mencari capacity factor dengan cara membagi energi aktual terhadap energi maksimum yang dihasilkan selama waktu 24 jam, Kemudian hasil tersebut dikalikan 100 agar menjadi persentase. Untuk menampilkan hasil dari perhitungan tersebut kita menggunakan f-string karena lebih mudah dibaca.

Dengan nilai energi harian yang diperoleh adalah 3600 kWh, nilai capacity factor yang didapat adalah 75,00%.

Diketahui data tekanan uap (bar) pada 5 pembacaan turbin:

tekanan = [12.4, 13.1, 12.8, 14.2, 13.5]

c. Tambahkan dua pembacaan baru, 13.9 lalu 12.6, ke dalam list. Tampilkan jumlah data setelah penambahan.

6. Status Operasi Boiler

Status sebuah boiler ditentukan dari suhu uap (°C) dan tekanan (bar):

"STATUS: TRIP" jika suhu mencapai 550 atau lebih, atau tekanan mencapai 14 atau lebih.

"STATUS: WARNING" jika suhu lebih dari 520 dan tekanan lebih dari 12.
selain itu,

"STATUS: NORMAL".

Baca suhu dan tekanan dari input (bertipe float).

Gunakan if-elif-else dengan operator logika. Perhatikan urutan pemeriksaan kondisi.

Uji program dengan kombinasi berikut:

suhu	tekanan	Output yang diharapkan

550	10	STATUS: TRIP

540	14	STATUS: TRIP

530	13	STATUS: WARNING

530	11	STATUS: NORMAL

520	13	STATUS: NORMAL


```python
#Status Operasi Boiler

suhu = float(input("Masukkan suhu (C): "))
tekanan = float(input("Masukkan tekanan (bar): "))

if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")

elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")

else:
    print("STATUS: NORMAL")
```

Pada program 6 kita menentukan status operasi boiler berdasarkan nilai suhu dan tekanan, pengguna diminta memasukkan suhu dalam satuan Celsius dan tekanan dalam satuan bar.

Perintah float() digunakan agar nilai yang dimasukkan dapat berupa angka desimal, terdapat 3 kondisi dimana if digunakan untuk memeriksa apakah suhu mencapai 550°C atau tekanan mencapai 14 bar, jika terpenuhi maka statusnya TRIP. Selanjutnya, elif digunakan untuk memeriksa apakah suhu lebih dari 520°C dan tekanan lebih dari 12 bar, jika terpenuhi maka statusnya WARNING.

Jika kedua kondisi tersebut tidak terpenuhi, perintah else akan menampilkan status NORMAL.

8. Pengosongan Tangki Bahan Bakar

Simulasikan pengosongan tangki dari 100 liter hingga kosong.
Mulai dengan volume = 100.

Gunakan perulangan while: selama volume > 0, kurangi 20 liter,

lalu cetak volume saat ini (setelah dikurangi).

Saat volume mencapai tepat 20 liter, cetak "WARNING: Hampir kosong!".

Setelah loop selesai, cetak pesan tangki kosong.

Contoh output:

Volume: 80 liter

Volume: 60 liter

Volume: 40 liter

Volume: 20 liter

WARNING: Hampir kosong!

Volume: 0 liter

TANGKI KOSONG - Pengosongan berhenti.

```python
#Pengosongan Tangki Bahan Bakar

volume = 100

while volume > 0:
    volume -= 20
    print(f"VOLUME: {volume} L")

    if volume == 20:
        print("WARNING: Hampir kosong!")

print("TANGKI KOSONG - Pengosongan berhenti.")
```

Pada program 8 kita di perintahkan  untuk menentukan status operasi boiler berdasarkan suhu dan tekanan yang dimasukkan oleh pengguna.

perintah input() digunakan untuk memasukkan nilai suhu dan tekanan, sedangkan pada float() mengubah input menjadi angka desimal, pada kondisi pertama, if digunakan untuk memeriksa apakah suhu mencapai 550°C atau tekanan mencapai 14 bar, jika terpenuhi maka statusnya TRIP, jika tidak, kondisi elif memeriksa apakah suhu lebih dari 520°C dan tekanan lebih dari 12 bar, jika terpenuhi maka statusnya WARNING. Apabila kedua kondisi tersebut tidak terpenuhi, program akan menampilkan status NORMAL dengan tersrbut program mengetahui kondisinya

10. Pembacaan Suhu Bearing dengan Sensor Rusak

Data suhu bearing lima titik pengukuran dibaca sebagai teks, dan satu sensor rusak mengirim "N/A":

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

Periksa setiap data dengan for. Nomor titik dimulai dari 1.

a. Ubah data menjadi float dengan try-except ValueError. Jika gagal, cetak Titik [n]: data tidak valid - pemeriksaan dihentikan, lalu hentikan perulangan dengan break.

b. Jika nilai melebihi 70.0, cetak Warning: Titik [n] suhu bearing tinggi!.

c. Hitung jumlah data valid yang berhasil diperiksa, dan tampilkan setelah perulangan selesai.

Output yang diharapkan untuk data di atas:

Warning: Titik 3 suhu bearing tinggi!
Titik 4: data tidak valid - pemeriksaan dihentikan
Jumlah data valid: 3

```python
# Pembacaan Suhu Bearing dengan Sensor Rusak

data =  ["62.5", "58.0", "81.3", "N/A", "66.0"]

# a. Ubah data menjadi float dengan try-except ValueError. Jika gagal,
# cetak Titik [n]: data tidak valid - pemeriksaan dihentikan,
# lalu hentikan perulangan dengan break.

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

for titik, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
        jumlah_valid += 1

        if suhu > 70.0:
            print(f"Warning: Titik {titik} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {titik}: data tidak valid - pemeriksaan dihentikan")
        break

print(f"Jumlah data valid: {jumlah_valid}")
```

Pada program 10 ini kita diperintahkan untuk memeriksa data suhu bearing menggunakan bahasa Python, terdapat lima nilai suhu, tetapi terdapat satu data tidak valid, yaitu "N/A".

Kita menggunakan perintah for untuk membaca data satu per satu, sedangkan enumerate() digunakan untuk memberikan nomor pada setiap titik pembacaan. Perintah try digunakan untuk mengubah data menjadi tipe angka desimal menggunakan float().

Jika suhu lebih dari 70°C, program akan menampilkan peringatan bahwa suhu bearing tinggi dan apabila ditemukan data yang tidak valid dalam program except ValueError akan menampilkan pesan bahwa pemeriksaan dihentikan, kemudian break digunakan untuk menghentikan perulangan untuk mengetahui jumlah_valid dalan program kita menghitung jumlah data yang berhasil dibaca, yaitu sebanyak 3 data valid.
