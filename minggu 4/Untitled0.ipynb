```python
#Hands on Pratikum

#1
total = float(input("Jarak tenpu: "))
Nama = str(input("Titik akhir: "))
Nama = str(input("Titik awal : "))

#2
Nama_tempat = "POLTEK"
Jarak_KM = 18.2
Kecepatan_KM_JAM = 60
Jumlah_tumpangan= 12

print("Bus gerak dengan kecepatan: %d" % (Kecepatan_KM_JAM))
print("Dengan jarak tempu: %f" % (Jarak_KM))
print("Menuju: %s" % (Nama_tempat))
print("Jumlah tumpangan: %x" % (Jumlah_tumpangan))
print("Tambahan penumpang: %o" % (Jumlah_tumpangan))

#3
jumlah_detik = int(input("Masukkan nilai"))

jam = jumlah_detik // 3600
sisa_detik = jumlah_detik % 3600
menit = sisa_detik // 6
detik = sisa_detik % 60

print("Hasil: %d jam, %d menit, %d detik" % (jam, menit, detik))

#4
tugas = float(input("Masukkan nilai Tugas: "))
uts = float(input("Masukkan nilai UTS: "))
uas = float(input("Masukkan nilai UAS: "))

nilai_akhir = (0.3 * tugas) + (0.3 * uts) + (0.4 * uas)

nilai_mahasiswa = nilai_akhir

standar_kelulusan = float(input("Masukkan standar kelulusan: "))

apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print("Nilai Akhir:", nilai_akhir)
print("Nilai Mahasiswa:", nilai_mahasiswa)
print("Standar Kelulusan:", standar_kelulusan)
print("Apakah Mahasiswa Lulus:", apakah_lulus)
print("Perlu Remedial:", perlu_remedial)

#5
belanja = float(input("Masukkan jumlah belanja: "))
diskon = (0, 0.35 * belanja) [belanja > 1000000]
total_harga = belanja - diskon

print("Total harga setelah diskon =", total_harga)
```
