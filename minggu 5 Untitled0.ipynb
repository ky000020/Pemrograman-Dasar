Budi membeli 7 buku dengan harga Rp 18.500 per buku. Toko memberi diskon 10% dari total belanja. Budi membayar dengan uang Rp 150.000.

```python
#NAMA  : HENGKY RYANTO SIHOMBING
#NIM   : 4232501055
#KELAS : TRPE PAGI_B

#1 Total belanja, besar diskon, dan jumlah yang harus dibayar.
nama_barang = "BUKU"
jumlah_barang = 7
harga_barang = 18500
bayar = 150000

seluruh_harga = jumlah_barang * harga_barang
diskon = 0.1 * seluruh_harga
total_harus_dibayar = seluruh_harga - diskon

print("SELURUH HARGA: ", seluruh_harga)
print("BESAR DISKON: ", diskon)
print("TOTAL HARUS DIBAYAR: ", total_harus_dibayar)

#2 Uang kembalian.
uang_kembalian = bayar - total_harus_dibayar
print("UANG KEMBALIAN: ", uang_kembalian)

#3 Jika jumlah yang dibayar dipatungi 4 orang secara adil dalam bilangan bulat,
# berapa iuran per orang dan berapa rupiah sisanya?
iuran_per_orang = int(total_harus_dibayar // 4)
sisanya = int(total_harus_dibayar % 4)

print("PATUNGAN PER ORANG: ", iuran_per_orang)
print("SISA PATUNGAN (Rp): ", sisanya)

#4 Apakah uang Budi cukup? (hasil True/False)
uang_cukup = bayar >= total_harus_dibayar
print("APAKAH UANG CUKUP: ", uang_cukup)
```

Sebuah rumah memasang 12 panel surya berdaya 450 Wp per panel. Rata-rata penyinaran efektif 5 jam/hari dengan efisiensi sistem 80% (0,8). Kebutuhan listrik rumah 25 kWh/hari dan tarif listrik Rp 1.444,70/kWh.


```python
#1 Energi yang dihasilkan per hari (kWh)
energi_harian = (12 * 450 * 5 * 0.8) / 1000
kebutuhan_harian = 25

print("ENERGI HARIAN (kWh) : ", energi_harian)

#2 Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan.
energi_bulanan = energi_harian * 30
nilai_rupiah = energi_bulanan * 1444.70

print("ENERGI BULANAN (kWh) : ", energi_bulanan)
print("NILAI RUPIAH LISTRIK (Rp) : ", nilai_rupiah)

#3 Kekurangan energi per hari terhadap kebutuhan.
kekurangan_energi = kebutuhan_harian - energi_harian

print("KEKURANGAN HARIAN (kWh) : ", kekurangan_energi)

#4 Apakah PLTS sudah mencukupi kebutuhan rumah? (True/False)
mencukupi = energi_harian >= kebutuhan_harian
print("MENCUKUPI KEBUTUHAN : ", mencukupi)
```

Sebuah generator diesel 200 kW beroperasi 8 jam/hari. Konsumsi bahan bakar 0,25 liter per kWh, harga solar Rp 13.500/liter, dan kapasitas tangki 1.000 liter. Kabel penghantar dialiri arus 20 A dengan hambatan 0,5 Ω.

```python
#1 Energi yang dibangkitkan per hari (kWh)
daya_generator = 200
jam_operasi = 8
energi_harian = daya_generator * jam_operasi
print("1. ENERGI HARIAN (kWh) : ", energi_harian)

#2 Kebutuhan solar per hari (liter) dan biayanya (Rp)
konsumsi_per_kwh = 0.25
harga_solar = 13500
konsumsi_solar_harian = energi_harian * konsumsi_per_kwh
biaya_solar_harian = konsumsi_solar_harian * harga_solar
print("2. KONSUMSI SOLAR HARIAN (liter)   : ", konsumsi_solar_harian)
print("   BIAYA SOLAR HARIAN (Rp)        : ", biaya_solar_harian)

#3 Berapa hari penuh tangki cukup, dan sisa solar (liter)
kapasitas_tangki = 1000
hari_penuh = int(kapasitas_tangki // konsumsi_solar_harian)
sisa_solar = kapasitas_tangki % konsumsi_solar_harian
print("3. KETAHANAN TANGKI (Hari Penuh)   : ", hari_penuh)
print("   SISA SOLAR DI TANGKI (liter)    : ", sisa_solar)

#4 Rugi daya pada kabel (P = I² × R)
arus = 20
hambatan = 0.5
rugi_daya_watt = arus * arus * hambatan
print("4. RUGI DAYA KABEL (Watt)          : ", round(rugi_daya_watt, 2))

#5 Apakah rugi kabel tergolong tinggi? (> 150 W dan arus >= 20 A)
rugi_tinggi = (rugi_daya_watt > 150) and (arus >= 20)
print("5. RUGI KABEL TINGGI (True/False)  : ", rugi_tinggi)
```
